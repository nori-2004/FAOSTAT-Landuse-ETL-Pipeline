```python
import pandas as pd

crops_raw = pd.read_csv("../data/raw/raw_crops.csv")
value_raw = pd.read_csv("../data/raw/raw_production_value.csv")

for name, df in [("CROPS", crops_raw), ("VALUE", value_raw)]:
    print(f"\n{'='*60}")
    print(f"{name} SHAPE: {df.shape}")
    
    numeric_cols = df.select_dtypes(include="number").columns
    
    print(f"\n{name} SUMMARY STATS:")
    print(df[numeric_cols].describe())
    
    print(f"\n{name} NULL COUNTS:")
    print(df.isnull().sum().sort_values(ascending=False))
```

```python
crops_raw = crops_raw.drop(columns=["Note"])

null_rows_crops = crops_raw[crops_raw.isnull().any(axis=1)]
null_rows_value = value_raw[value_raw.isnull().any(axis=1)]

print("CROPS rows with nulls:")
print(null_rows_crops)

print("\nVALUE rows with nulls:")
print(null_rows_value)
```
